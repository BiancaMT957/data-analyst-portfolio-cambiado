```python
import sqlite3
import pandas as pd

conexion = sqlite3.connect("../bases_datos/ventas.db")
```

Consultas con join

```python
consulta = """
    SELECT v.IdVenta,
           c.NombreCliente,  
           v.FechaVenta
    FROM Ventas v
    INNER JOIN Clientes c
      ON v.IdCliente = c.IdCliente
"""
ventas = pd.read_sql_query(consulta, conexion)
ventas.head()
```
