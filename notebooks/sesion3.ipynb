```python
Importar librerias para conectar a sqlite y poder hacer consultas
```

```python
import sqlite3
import pandas as pd
```

Creamos la conexion a la base de datos

```python
conexion= sqlite3.connect("../bases_datos/ventas.db")
```

Vamos a crear las consultas que necesite

```python
consultaClientesAlicante= """
SELECT NombreCliente, Ciudad, Provincia
FROM Clientes
WHERE Provincia= 'Alicante'
"""
consultaVentas1T = """
SELECT * FROM Ventas
WHERE FechaVenta BETWEEN '2025-01-01' AND '2025-03-31'
"""
#Precio entre 50 y 150
consultaPrecioIntervalo= """
SELECT NombreProducto, PrecioVenta
FROM Productos
WHERE PrecioVenta BETWEEN 50 AND 150;
"""
```

Ahora usamos la libreria pandas para hacer las consultas y guardarlas en objetos DataFrame de Python

```python
dfClientesAlicante = pd.read_sql_query(consultaClientesAlicante, conexion)
dfVentas1T = pd.read_sql_query(consultaVentas1T, conexion)
dfPrecioIntervalo = pd.read_sql_query(consultaPrecioIntervalo, conexion)
```

```python
print("ClientesAlicante:")
print(dfClientesAlicante)
print()
print("Ventas1T:")
print(dfVentas1T)
print()
print("PrecioIntervalo:")
print(dfPrecioIntervalo)
```
