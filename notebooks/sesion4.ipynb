Importamos las librerias de nuetra base de datos

```python
import sqlite3
import pandas as pd
```

Creamos la conexion a nuetra nase de datos. Usamos sqlite3

```python
conexion = sqlite3.connect("../bases_datos/ventas.db")
```

Creamos variables que guarden el codigo de nuestas consultas SQL. (Usamos las 3" para poder tabular la consulta)

```python
consultaNumClientes = """
SELECT COUNT(*) As NumClientes FROM Clientes;
"""
consultaPVPMedio = """
SELECT AVG(PrecioVenta) As PrecioMedio FROM Productos;
"""
consultaDatosVentaEnero = """
SELECT SUM(Cantidad*PrecioUnitario) As ImporteVentas,
AVG(Cantidad*PrecioUnitario) As TicketMedio
FROM LineasVenta, Ventas
WHERE LineasVenta.IdVenta = Ventas.IdVenta AND
FechaVenta BETWEEN "2025-01-01" AND "2025-01-31";
"""
```

Vamos a crear los dataframes para manejar los resultados de las consultas en Python. Para el utilizaremos Pandas que hemos renombrado pd.

```python
dfNumClientes = pd.read_sql_query(consultaNumClientes, conexion)
dfPVPMedio = pd.read_sql_query(consultaPVPMedio, conexion)
dfDatosVentaEnero = pd.read_sql_query(consultaDatosVentaEnero, conexion)
```

Podemos seguir trabajando desde Python con los dataframe que hemos creado

```python
print("Numero de Clientes:")
print(dfNumClientes)
print()
print("Precio Medio de los productos:")
print(dfPVPMedio)
print()
print("Datos de Venta de Enero 2025:")
print(dfDatosVentaEnero)
```
