```python
import sqlite3
import pandas as pd
```

Creamos conexión a nuestra base de datos

```python
conexion = sqlite3.connect("../bases_datos/ventas.db")
```

Escribimos el código SQL de nuestra consulta(s) asignándolos a una variable

```python
consultaPrecioMedio = """
SELECT IdCategoria,
AVG(PrecioVenta) AS PrecioMedio
FROM Productos
GROUP BY IdCategoria
HAVING PrecioMedio > 100
ORDER BY PrecioMedio DESC;
"""
```

Ahora usamos pandas para pasar la consulta a la base de datos y traernos el resultado a un objeto dataframe de python

```python
dfPrecio = pd.read_sql_query(consultaPrecioMedio, conexion)
```

Si ya henos termiando de preguntar a la base de datos, cerramos la conexión

```python
conexion.close()
```
