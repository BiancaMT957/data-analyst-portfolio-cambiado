Guardamos la consulta corregida como texto en Python

```python
import sqlite3
import pandas as pd
import matplotlib.pyplot as pl
conexion = sqlite3.connect("../bases_datos/ventas.db")
```

```python
consulta = """
SELECT v.FechaVenta, c.NombreCliente
FROM Ventas v
INNER JOIN Clientes c ON v.IdCliente = c.IdCliente;
"""
```

La ejecutamos con pandas usando la conexión activa

```python
ventas = pd.read_sql_query(consulta, conexion)
```
