Aqui guardamos el codig SQL de la consulta en este caso usaremos una vista (vRangoPrecios)

```python
conRangoPrecios = """
SELECT * FROM vRangoPrecios;
"""
conProductosMayorMedia="""
SELECT NombreProducto, PrecioVenta
FROM Productos
WHERE PrecioVenta>(SELECT AVG(PrecioVenta) FROM Productos;
"""
conPrecioMenorMedia ="""
SELECT NombreProducto, PrecioVenta
FROM Productos
WHERE PrecioVenta>(SELECT AVG(PrecioVenta) FROM Productos;
"""
conCLientesCompra = """
SELECT * FROM Clientes
WHERE IdCliente IN(SELECT DISTINCT IdCliente FROM Ventas;
"""
conClientesNOCompra = """
SELECT * FROM Clientes
WHERE IdCliente IN(SELECT DISTINCT IdCliente FROM Ventas;
"""
conClasificacionProductos ="""
SELECT NombreProducto,
PrecioVenta,
CASE
WHEN PrecioVenta< 20 THEN'Económico'
WHEN PrecioVenta< 50 THEN'Estandar'
ELSE AS TipoProducto,
Stock, 
CASE
WHEN Stock <v10 THEN 'Critico'
WHEN Stock <50 THEN 'Medio'
END AS NivelStock
FROM Productos;
"""
```

Ejecutamos la consulta a la vista para traerlo a un dataframe

```python
dfRangoPrecios = pd.read_sql_query(conRangoPrecios, conexion)
dfProductosMayorMedia = pd.sql_query(conProductosMayorMedia, conexion)
dfPrecioMenorMedia = pd.sql_query(conPrecioMenorMedia, conexion)
dfCLientesCompra = pd.sql_query(conCLientesCompra, conexion)
dfClientesNOCompra = pd.sql_query(conClientesNOCompra , conexion)
dfClasificacionProductos = pd.sql_query(conClasificacionProductos, conexion)
```
