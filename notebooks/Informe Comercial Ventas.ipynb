```python
import sqlite3
import pandas as pd

conexion = sqlite3.connect("../bases_datos/ventas.db")
```

```python
consulta_clientes = "SELECT COUNT(*) AS TotalClientes FROM Clientes;"
pd.read_sql_query(consulta_clientes, conexion)
```

```python
consulta_productos = """
SELECT
    COUNT(*) AS TotalProductos,
    AVG(PrecioVenta) AS PrecioMedio,
    MIN(PrecioVenta) AS PrecioMinimo,
    MAX(PrecioVenta) AS PrecioMaximo,
    SUM(Stock) AS StockTotal
FROM Productos;
"""
pd.read_sql_query(consulta_productos, conexion)
```

```python
consulta_ventas = """
SELECT
    COUNT(DISTINCT Ventas.IdVenta) AS TotalVentasRegistradas,
    SUM((Cantidad * PrecioUnitario) - ((Cantidad * PrecioUnitario * (Descuento / 100)))) AS FacturacionTotal
FROM LineasVenta
INNER JOIN Ventas ON LineasVenta.IdVenta = Ventas.IdVenta;
"""
pd.read_sql_query(consulta_ventas, conexion)
```
